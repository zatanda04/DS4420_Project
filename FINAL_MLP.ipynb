```python
# libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings("ignore")
```

```python
# loading files
fem = pd.read_csv("NSFG_2022_2023_FemRespPUFData.csv")
male = pd.read_csv("NSFG_2022_2023_MaleRespPUFData.csv")

print(fem.shape)
print(male.shape)
```

## **Cleaning**

We will synthesize the datasets and add a sex column. Once filtered for divorved individuals, we will further extract the data based on the socioeconomic factors used to conduct the analysis.

```python
# Extracting female and male divorcees
fem_div  = fem[fem["MARENDHX_1"] == 2].copy()
male_div = male[male["MARREND_W1"] == 2].copy()

# FMARNO counts marriages, if someone is divorced and FMARNO > 1, they have remarried
male_div["remarried"] = (male_div["FMARNO"] > 1).astype(int)
fem_div["remarried"]  = (fem_div["FMARNO"] > 1).astype(int)

print(f"\nFemale divorced: {len(fem_div)}  |  remarried: {fem_div['remarried'].sum()}")
print(f"Male   divorced: {len(male_div)}  |  remarried: {male_div['remarried'].sum()}")
```

```python
# pulling cols to keep
factors  = ["AGE_R", "POVERTY", "TOTINCR", "HIEDUC", "HISP", "LABORFOR", "RSCRRACE", "remarried",]

# changing col name to num_children
fem_div["num_children"] = fem_div["PARITY"]
male_div["num_children"] = male_div["NUMBIOKID"]

fem_div["sex"]  = 1
male_div["sex"] = 0

# combining to one common column list, then into a single dataset
keep = factors + ["num_children", "sex"]
remarriage_data = pd.concat([fem_div[keep], male_div[keep]], ignore_index=True)

# standardizing col names
remarriage_data = remarriage_data.rename(columns={
    "AGE_R"    : "age",
    "POVERTY"  : "poverty_ratio",
    "TOTINCR"  : "total_income_cat",
    "HIEDUC"   : "education_level",
    "LABORFOR" : "labor_force_status",
    "RSCRRACE" : "race",
    "HISP"     : "hispanic"
})

print(f"\nCombined dataset: {remarriage_data.shape[0]} rows, {remarriage_data.shape[1]} columns")
print(remarriage_data.columns)
```

```python
# replacing NFSG codes with NAN
remarriage_data["hispanic"] = remarriage_data["hispanic"].replace({8: np.nan, 9: np.nan})
remarriage_data["num_children"] = remarriage_data["num_children"].replace({98: np.nan, 99: np.nan})
remarriage_data["age"] = remarriage_data["age"].where(remarriage_data["age"] < 95, np.nan)
remarriage_data["poverty_ratio"] = remarriage_data["poverty_ratio"].where(remarriage_data["poverty_ratio"] <= 700, np.nan)
remarriage_data["race"] = remarriage_data["race"].replace({8: np.nan, 9: np.nan})


# recoding categorical variables

# hispanic - binary: 1 = Hispanic, 0 = not Hispanic
remarriage_data["hispanic"] = remarriage_data["hispanic"].map({1: 1, 5: 0})

# labor_force_status - binary: employed (codes 1–3) vs. not employed (codes 4–7)
remarriage_data["employed"] = remarriage_data["labor_force_status"].apply(
    lambda x: 1 if x in [1, 2, 3] else (0 if x in [4, 5, 6, 7] else np.nan)
)
remarriage_data = remarriage_data.drop(columns=["labor_force_status"])

# race - one-hot encode (White = reference category, dropped)
race_dummies = pd.get_dummies(remarriage_data["race"], prefix="race", dummy_na=False)
race_dummies = race_dummies.rename(columns={
    "race_1": "race_other",
    "race_2": "race_black",
    "race_3": "race_white",
    "race_4": "race_hispanic"
})

# drop white as the reference category (largest group)
race_dummies = race_dummies.drop(columns=["race_white"], errors="ignore")
remarriage_data = pd.concat([remarriage_data.drop(columns=["race"]), race_dummies],axis=1)

# creating bins for income and education level
# Income - low, middle, high
remarriage_data["income_group"] = pd.cut(
    remarriage_data["total_income_cat"],
    bins=[0, 5, 10, 15],
    labels=[0, 1, 2]
)

# Education - low, medium, high
remarriage_data["education_group"] = pd.cut(
    remarriage_data["education_level"],
    bins=[0, 4, 8, 11],
    labels=[0, 1, 2]
)
# dropping original col and converting new ones
remarriage_data = remarriage_data.drop(columns=["total_income_cat", "education_level"])
remarriage_data["income_group"] = remarriage_data["income_group"].astype(int)
remarriage_data["education_group"] = remarriage_data["education_group"].astype(int)
# imputing remaining NAN with median
for col in remarriage_data.columns:
    if remarriage_data[col].isna().any():
        remarriage_data[col] = remarriage_data[col].fillna(remarriage_data[col].median())

# check
assert remarriage_data.isna().sum().sum() == 0, "Still NaNs"
print("\nCleaning complete")
print(f"Remarried: {remarriage_data['remarried'].mean():.1%} | Not: {1 - remarriage_data['remarried'].mean():.1%}")
print(f"Final feature columns: {[c for c in remarriage_data.columns if c != 'remarried']}")
remarriage_data.head()
```

## **Preprocessing for Multilayer Perceptron (MLP)**
We are splitting, testing, and training the data into their respective sets while keeping the split of two class––remarried and not remarried

```python
# features
features = [col for col in remarriage_data.columns if col != "remarried"]

X = remarriage_data[features].values.astype(float)
y = remarriage_data["remarried"].values.astype(float)
```

```python
# standardizing parameters
def standardize(X_train, X_test):
    mu  = X_train.mean(axis=0)
    std = X_train.std(axis=0)
    std[std == 0] = 1
    return (X_train - mu) / std, (X_test - mu) / std

# train, test, and split
np.random.seed(42)
pos_idx = np.where(y == 1)[0]
neg_idx = np.where(y == 0)[0]

# spliting into remarried and not remarried classes
def split_class(idx, frac=0.8):
    np.random.shuffle(idx)
    c = int(len(idx) * frac)
    return idx[:c], idx[c:]

pos_train, pos_test = split_class(pos_idx)
neg_train, neg_test = split_class(neg_idx)

train_idx = np.concatenate([pos_train, neg_train])
test_idx  = np.concatenate([pos_test,  neg_test])

X_train, y_train = X[train_idx], y[train_idx]
X_test,  y_test  = X[test_idx],  y[test_idx]

X_train, X_test = standardize(X_train, X_test)
print(f"\nTraining set size: {len(X_train)}  |  Testing set size: {len(X_test)}")
```

## **Multilayer Perceptron (MLP) Neural Network**
We will now implement a fully manual MLP using the parameters:

- **Activations:** ReLU (hidden layers), Sigmoid (output)
- **Loss:** Binary Cross-Entropy
- **Optimizer:** Mini-batch SGD
- **Architecture search:** 1–3 hidden layers, widths 32–128

```python
# activation functions
def relu(z):
    return np.maximum(0, z)

def relu_derivative(z):
    return (z > 0).astype(float)

def sigmoid(z):
    z = np.clip(z, -500, 500)
    return 1 / (1 + np.exp(-z))

# loss function
def binary_cross_entropy(y_true, y_pred):
    y_pred = np.clip(y_pred, 1e-9, 1 - 1e-9)
    return -np.mean(y_true * np.log(y_pred) + (1 - y_true) * np.log(1 - y_pred))
```

```python
# training func for the MLP to use different architectures
def train_mlp(hidden_layers):
    np.random.seed(42)
    layer_sizes = [N_INPUT] + hidden_layers + [N_OUTPUT]

    weights = [np.random.randn(layer_sizes[i], layer_sizes[i+1]) * np.sqrt(2.0 / layer_sizes[i])
               for i in range(len(layer_sizes) - 1)]
    biases  = [np.zeros((1, layer_sizes[i+1]))
               for i in range(len(layer_sizes) - 1)]

    loss_history = []

    for epoch in range(EPOCHS):

        # shuffling training data each epoch
        perm = np.random.permutation(len(X_train))
        X_shuf, y_shuf = X_train[perm], y_train[perm]

        epoch_losses = []

        for start in range(0, len(X_train), BATCH_SIZE):
            Xb = X_shuf[start : start + BATCH_SIZE]
            yb = y_shuf[start : start + BATCH_SIZE, None]

            # forward pass
            activations, pre_acts = [Xb], []
            A = Xb
            for i, (W, b) in enumerate(zip(weights, biases)):
                Z = A @ W + b
                pre_acts.append(Z)
                A = sigmoid(Z) if i == len(weights) - 1 else relu(Z)
                activations.append(A)

            loss = binary_cross_entropy(yb, activations[-1])
            epoch_losses.append(loss)

            # backward pass
            B = Xb.shape[0]
            dA = (activations[-1] - yb) / B

            for i in reversed(range(len(weights))):
                dZ = dA if i == len(weights) - 1 else dA * relu_derivative(pre_acts[i])
                dW = activations[i].T @ dZ
                db = dZ.sum(axis=0, keepdims=True)
                dA = dZ @ weights[i].T

                # update weights
                weights[i] -= LR * dW
                biases[i]  -= LR * db

        loss_history.append(np.mean(epoch_losses))

        if (epoch + 1) % 50 == 0:
            print(f"Epoch {epoch+1:>3}/{EPOCHS}  |  Loss: {loss_history[-1]:.4f}")

    # forward pass on test
    A = X_test
    for i, (W, b) in enumerate(zip(weights, biases)):
        Z = A @ W + b
        A = sigmoid(Z) if i == len(weights) - 1 else relu(Z)

    y_proba = A.flatten()
    y_pred = (y_proba >= 0.5).astype(int)

    return weights, biases, loss_history, y_proba, y_pred
```

```python
# hyperparameters
N_INPUT    = X_train.shape[1]
N_OUTPUT   = 1
LR         = 0.01
EPOCHS     = 200
BATCH_SIZE = 32

# architectures - 1,2,3 layers
architectures = {
    "1-layer  [32]": [32],
    "1-layer  [64]": [64],
    "2-layers [64, 32]": [64, 32],
    "2-layers [128, 64]": [128, 64],
    "3-layers [128, 64, 32]": [128, 64, 32],
}

# finding the best architecture
best_acc = -1
for label, hidden_layers in architectures.items():
    print(f"\n--- {label} ---")
    weights, biases, loss_history, y_proba, y_pred = train_mlp(hidden_layers)
    acc = np.mean(y_pred == y_test.astype(int))
    print(f"Test Accuracy: {acc:.4f}")

    if acc > best_acc:
        best_acc = acc
        best_label = label
        best_weights = weights
        best_biases  = biases
        best_loss_history = loss_history
        best_y_proba = y_proba
        best_y_pred  = y_pred

# final variables
loss_history = best_loss_history
y_proba      = best_y_proba
y_pred       = best_y_pred

print(f"\nBest: {best_label} | test accuracy = {best_acc:.4f}")
```

```python
# evaluating metrics

def confusion(y_true, y_pred):
    TP = np.sum((y_pred == 1) & (y_true == 1))
    TN = np.sum((y_pred == 0) & (y_true == 0))
    FP = np.sum((y_pred == 1) & (y_true == 0))
    FN = np.sum((y_pred == 0) & (y_true == 1))
    return TP, TN, FP, FN

def classification_metrics(y_true, y_pred):
    TP, TN, FP, FN = confusion(y_true.astype(int), y_pred.astype(int))
    accuracy = (TP + TN) / len(y_true)
    precision = TP / (TP + FP) if (TP + FP) > 0 else 0
    recall = TP / (TP + FN) if (TP + FN) > 0 else 0
    f1 = 2 * precision * recall / (precision + recall) if (precision + recall) > 0 else 0
    return accuracy, precision, recall, f1, TP, TN, FP, FN

def roc_auc(y_true, y_scores):
    thresholds = np.sort(np.unique(y_scores))[::-1]
    tpr, fpr = [0.0], [0.0]

    P = np.sum(y_true)
    N = len(y_true) - P

    for t in thresholds:
        pred = (y_scores >= t).astype(int)
        TP = np.sum((pred == 1) & (y_true == 1))
        FP = np.sum((pred == 1) & (y_true == 0))
        tpr.append(TP / P if P > 0 else 0)
        fpr.append(FP / N if N > 0 else 0)

    tpr.append(1.0)
    fpr.append(1.0)

    auc = np.trapezoid(tpr, fpr)
    return auc, np.array(fpr), np.array(tpr)

accuracy, precision, recall, f1, TP, TN, FP, FN = classification_metrics(y_test, y_pred)
auc, fpr_pts, tpr_pts = roc_auc(y_test.astype(int), y_proba)
```

```python
# test results
print("\nTest Set Results")
print(f"Accuracy : {accuracy:.3f}")
print(f"Precision: {precision:.3f}")
print(f"Recall   : {recall:.3f}")
print(f"F1 Score : {f1:.3f}")
print(f"ROC-AUC  : {auc:.3f}")

print("\nConfusion Matrix:")
print(f"             Pred 0   Pred 1")
print(f"Actual 0:    {TN:>4}     {FP:>4}")
print(f"Actual 1:    {FN:>4}     {TP:>4}")
```

```python
# roc curve
plt.figure(figsize=(6, 5))
plt.plot(fpr_pts, tpr_pts, color="darkorange", linewidth=2,
         label=f"MLP (AUC = {auc:.3f})")
plt.plot([0, 1], [0, 1], "k--", linewidth=1, label="Random baseline")
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curve\nLikelihood of Remarriage After Divorce")
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# confusion matrix
cm = np.array([[TN, FP],
               [FN, TP]])

plt.figure(figsize=(5, 4))
plt.imshow(cm, cmap="Blues")
plt.colorbar()
plt.grid(False)
plt.xticks([0, 1], ["Predicted No", "Predicted Yes"])
plt.yticks([0, 1], ["Actual No", "Actual Yes"])
plt.title("Confusion Matrix\nLikelihood of Remarriage After Divorce")
for i in range(2):
    for j in range(2):
        if i == 1 and j == 0:
            text_color = "black"
        else:
            text_color = "black" if cm[i, j] > cm.max() / 2 else "black"
        plt.text(j, i, cm[i, j],
                 ha="center",
                 va="center",
                 fontsize=14,
                 color=text_color)
plt.tight_layout()
plt.show()
```

```python
# feature importance plot
baseline_acc = np.mean((y_proba >= 0.5).astype(int) == y_test.astype(int))
importance = []
np.random.seed(42)
# saves best weights/biases
for i in range(X_test.shape[1]):
    X_permuted = X_test.copy()
    np.random.shuffle(X_permuted[:, i])

    #  forward pass using saved best model
    A = X_permuted
    for j, (W, b) in enumerate(zip(best_weights, best_biases)):
        Z = A @ W + b
        A = sigmoid(Z) if j == len(best_weights) - 1 else relu(Z)
    perm_proba = A.flatten()

    perm_acc = np.mean((perm_proba >= 0.5).astype(int) == y_test.astype(int))
    importance.append(baseline_acc - perm_acc)

# sorting features by importance for a cleaner chart
importance     = np.array(importance)
sorted_idx     = np.argsort(importance)
sorted_names   = [features[i] for i in sorted_idx]
sorted_imp     = importance[sorted_idx]

plt.figure(figsize=(7, 5))
plt.barh(sorted_names, sorted_imp, color="mediumseagreen", edgecolor="darkgreen", linewidth=0.8)
plt.axvline(0, color="darkgreen", linewidth=0.8, linestyle="--")
plt.xlabel("Drop in Accuracy (higher = more important)")
plt.title("Permutation Feature Importance\nLikelihood of Remarriage After Divorce")
plt.tight_layout()
plt.show()
```

```python
# training loss curve
plt.figure(figsize=(6, 4))
plt.plot(range(1, EPOCHS + 1), loss_history, color="orange", linewidth=2)
plt.xlabel("Epoch")
plt.ylabel("Binary Cross-Entropy Loss")
plt.title("Training Loss Curve\nLikelihood of Remarriage After Divorce")
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```
